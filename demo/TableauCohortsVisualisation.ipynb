# Tableau Cohorts 

As well a chewing up the extracts as tableau dashboards it's fairly easy to do some basic visualizations in python. This notebook aims to demo that. 

```python
import pandas as pd
from pathlib import Path



# load files and apply filters 
path = Path.cwd()
file_path = path.parent / ".do_not_commit" / "ckd_demog_ukrdc_live_2024.csv"
df = pd.read_csv(file_path)

filters = {
    "adultpaed": "Adult",
    #"dialtplt": None,
    "country":"England",
    #"centre_code": "RAQ01",
    "centre_code": None,
    "satellite_code": None,
    #"quarter": None,
    #year": None,
    "region": None
}


# apply filters to the dataframe
filtered_df = df.copy()
for column, value in filters.items():
    if value is not None:
        filtered_df = filtered_df[filtered_df[column] == value]

filtered_df.head()
```

```python

import plotly.express as px

# plot each of the demographic measures over the most recent quarter
year_max = filtered_df['year'].max()
quarter_max = filtered_df[filtered_df['year'] == year_max]['quarter'].max()

# filter dataframe to most recent year and quarter
most_recent_demog = filtered_df[
    (filtered_df['year'] == year_max) 
    & (filtered_df['quarter'] == quarter_max)
]

# plot ethnicity
ethnicity_agg = most_recent_demog[
    most_recent_demog["variable2"] == "Ethnicity"
].groupby(["variable","dialtplt"])["value"].sum()


# Create stacked bar chart
fig = px.bar(
    ethnicity_agg.reset_index(),
    x="dialtplt",          # e.g. dialysis or transplant
    y="value",
    color="variable",      # e.g. ethnic categories
    title="Ethnicity Distribution",
    labels={"dialtplt": "Treatment Type", "value": "Count", "variable": "Ethnicity"},
    barmode="stack"
)

fig.show()
```

```python
# plot longitudinally
longitudinal_clinic = filtered_df[filtered_df['variable2'] == 'Clinic Type'].groupby(['variable','year','quarter'])["value"].sum().reset_index()


longitudinal_clinic["period"] = (
    longitudinal_clinic["year"].astype(str) + " Q" + longitudinal_clinic["quarter"].astype(str)
)



# Create stacked bar chart
fig = px.bar(
    longitudinal_clinic,
    x="period",
    y="value",
    color="variable",  # clinic type categories
    title="Clinic Type Distribution Over Time",
    labels={
        "period": "Year / Quarter",
        "value": "Count",
        "variable": "Clinic Type"
    },
    barmode="stack"
)

fig.update_layout(xaxis={'categoryorder':'category ascending'})
fig.show()
```
