# Demographics

This notebook drives the Demographics module for development and demonstration purposes. It assumes a ssh tunnel has been created to the required database and that a .env file exists somewhere with a filepath to the configuration settings and the database to connect to. 

```python
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.orm import sessionmaker
import json
import os

load_dotenv()


# set up connection to the database 
db_name = os.getenv("DB_NAME")
db_file = os.getenv("JSON_CONF_PATH")

with open(db_file) as file:
    conf = json.load(file)[db_name]
    db_url = f'postgresql+psycopg2://{conf["USER"]}:{conf["PASSWORD"]}@localhost:{conf["DB_PORT"]}/{conf["DATABASE"]}'


ukrdc3_sessionmaker = sessionmaker(
    autocommit=False,
    autoflush=False,
    bind=create_engine(db_url)
)

ukrdc3 = ukrdc3_sessionmaker()
```

Okay now we can drive the demographics library. 

```python
from dashboard_stats.demographics import Demog
import datetime as dt


ting = Demog(ukrdc3, 'RFBAK')
output = ting.patient_info()

print(output.json())
```
